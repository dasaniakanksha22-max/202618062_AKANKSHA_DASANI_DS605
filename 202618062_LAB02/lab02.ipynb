```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Task 1 

# Q1(a) Generate Array A with 100 random integers


np.random.seed(42)  

A = np.random.randint(1, 101, 100)

print("Q1(a) Array A (first 10 values):")
print(A[:10])


print("Minimum:", np.min(A))
print("Maximum:", np.max(A))
print("Median:", np.median(A))
print("Mean:", np.mean(A))
print("Standard Deviation:", np.std(A))

print("\n---\n")

# Q1(b) Generate Array B using np.arange()


B = np.arange(1, 101)

print("Q1(b) Array B (first 10 values):")
print(B[:10])

print("\n---\n")

# Q1(c) Implement np.zeros() and np.ones()


zeros_array = np.zeros(5)
ones_array = np.ones(5)

print("Q1(c)")
print("Zeros array:", zeros_array)
print("Ones array:", ones_array)

print("Shape of zeros array:", zeros_array.shape)
print("Data type of zeros array:", zeros_array.dtype)

print("\n---\n")


```

```python
# Q1(d) Implement np.linspace()


lin = np.linspace(0, 10, 5)

print("Q1(d) np.linspace():")
print(lin)

print("Difference from np.arange():")
print("np.arange() uses a step size, while np.linspace() creates a fixed number of equally spaced values.")

print("\n---\n")
```

```python
# Q1(e) Create one 2D and one 3D array


arr2d = np.arange(1, 13).reshape(3, 4)

print("Q1(e) 2D Array:")
print(arr2d)

print("Shape:", arr2d.shape)
print("ndim:", arr2d.ndim)

# Indexing
print("Element at row 2, column 3:", arr2d[1, 2])

# Rows and columns
print("First row:", arr2d[0])
print("Second column:", arr2d[:, 1])

# Slicing
print("Top-left 2x2 slice:")
print(arr2d[:2, :2])

print("\n---\n")

# 3D Array
arr3d = np.arange(24).reshape(2, 3, 4)

print("3D Array:")
print(arr3d)

print("Shape:", arr3d.shape)
print("ndim:", arr3d.ndim)

# Indexing and slicing
print("Element [1,2,3]:", arr3d[1, 2, 3])

print("First block:")
print(arr3d[0])

print("Slice arr3d[:, :2, :2]:")
print(arr3d[:, :2, :2])

print("\n---\n")
```

```python

# Task 2 

# Q2(a) Create two matrices


A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("\n---\n")

# Q2(b) Matrix addition


print("Addition (A + B):")
print(A + B)

print("\n---\n")


# Q2(c) Element-wise multiplication


print("Element-wise Multiplication (A * B):")
print(A * B)

print("\n---\n")

```

```python
# Q2(d) Matrix multiplication


print("Matrix Multiplication (A @ B):")
print(A @ B)

print("\n---\n")

# Q2(e) Transpose and Determinant


print("Transpose of A:")
print(A.T)

print("\nDeterminant of A:")
print(np.linalg.det(A))

print("\n---\n")


```

```python
# Q2(f) Inverse and Verification


inverse_A = np.linalg.inv(A)

print("Inverse of A:")
print(inverse_A)

print("\nVerification A @ A^-1:")
print(A @ inverse_A)

print("\nIs it approximately identity matrix?")
print(np.allclose(A @ inverse_A, np.eye(2)))

print("\n---\n")

# Q2(g) Vectorized NumPy operations (no loops)


vector = np.arange(1, 11)

print("Original vector:", vector)

print("Square of vector:", vector**2)

print("Sine of vector:", np.sin(vector))

print("Vector + 10:", vector + 10)

```
