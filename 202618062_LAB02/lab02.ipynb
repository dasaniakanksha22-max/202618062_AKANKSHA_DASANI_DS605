```python

# Task 4 


df = pd.read_csv("train.csv")

print("First 5 rows:")
display(df.head())

print("\n---\n")

print("Last 5 rows:")
display(df.tail())

print("\n---\n")

print("Shape of dataset:", df.shape)

print("\n---\n")

print("Columns:")
print(df.columns.tolist())

print("\n---\n")

print("Dataset Info:")
df.info()

print("\n---\n")


print("Statistical Summary:")
display(df.describe())
```

```python

# Task 5 

# 1. Male passengers older than 50
male_over_50 = df[(df["Sex"] == "male") & (df["Age"] > 50)]

print("1. Male passengers older than 50:", len(male_over_50))

print("\n---\n")

# 2. Female first-class passengers and survival percentage
female_first = df[(df["Sex"] == "female") & (df["Pclass"] == 1)]

count_ff = len(female_first)
survival_ff = female_first["Survived"].mean() * 100

print("2. Female first-class passengers:", count_ff)
print("Survival percentage:", round(survival_ff, 2), "%")

print("\n---\n")

# 3. Age 20-40, Fare above overall median, and survived
fare_median = df["Fare"].median()

filtered_20_40 = df[
    (df["Age"].between(20, 40)) &
    (df["Fare"] > fare_median) &
    (df["Survived"] == 1)
]

print("3. Age 20-40, Fare above median, and survived:", len(filtered_20_40))

print("\n---\n")

# 4. Traveling alone, age < 30, and did not survive
alone_not_survived = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

print("4. Traveling alone, age < 30, and did not survive:", len(alone_not_survived))

print("\n---\n")

# 5. Embarked='S', Pclass 2 or 3, and Fare above Southampton median
southampton_median = df[df["Embarked"] == "S"]["Fare"].median()

south_filtered = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median)
]

print("5. Embarked='S', Pclass 2 or 3, Fare above Southampton median:", len(south_filtered))
```

```python

# Task 6 

# 1. Average survival rate by Sex
print("1. Average Survival Rate by Sex")
display(df.groupby("Sex")["Survived"].mean())

print("\n---\n")

# 2. Average survival rate by Pclass
print("2. Average Survival Rate by Pclass")
display(df.groupby("Pclass")["Survived"].mean())

print("\n---\n")

# 3. Mean Age and Fare by Pclass
print("3. Mean Age and Fare by Pclass")
display(df.groupby("Pclass")[["Age", "Fare"]].mean())

print("\n---\n")

# 4. Multi-level grouping by Sex and Pclass
print("4. Multi-level Grouping (Sex and Pclass)")
multi_group = df.groupby(["Sex", "Pclass"])["Survived"].agg(["count", "mean"])

display(multi_group)

print("\n---\n")

# 5. Embarked-wise Fare and Survival statistics
print("5. Embarked-wise Fare and Survival Statistics")

embarked_stats = df.groupby("Embarked")[["Fare", "Survived"]].agg(["count", "mean"])

display(embarked_stats)
```

```python

# Task 7 

# 1. Count missing values
print("1. Missing Values Count")

missing_count = df.isnull().sum()
display(missing_count)

print("\n---\n")

# 2. Percentage of missing values
print("2. Missing Values Percentage")

missing_percent = (missing_count / len(df)) * 100
display(missing_percent.round(2))

print("\n---\n")

# 3. Visualize missing values
print("3. Missing Values Visualization")

plt.figure(figsize=(8, 4))
missing_count.plot(kind="bar")

plt.title("Missing Values in Titanic Dataset")
plt.xlabel("Columns")
plt.ylabel("Count")

plt.show()

print("\n---\n")

# 4. Fill Age with mean
print("4. Fill Age with Mean")

before_missing = df["Age"].isnull().sum()

df["Age"] = df["Age"].fillna(df["Age"].mean())

after_missing = df["Age"].isnull().sum()

print("Missing Age values before:", before_missing)
print("Missing Age values after:", after_missing)

print("\n---\n")

# 5. Median Age and Mode Embarked
print("5. Median Age and Mode Embarked")

median_age = df["Age"].median()
mode_embarked = df["Embarked"].mode()[0]

print("Median Age:", median_age)
print("Mode Embarked:", mode_embarked)

print("\n---\n")

# 6. Detect Fare outliers using IQR
print("6. Fare Outlier Detection using IQR")

Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df["Fare"] < lower_bound) | (df["Fare"] > upper_bound)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Fare Outliers:", len(outliers))
```

```python

# Task 8 - Feature Engineering and Pivot Table


# 1. Create FamilySize
print("1. Creating FamilySize")

df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

display(df[["SibSp", "Parch", "FamilySize"]].head())

print("\n---\n")

# 2. Create IsAlone
print("2. Creating IsAlone")

df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

display(df[["FamilySize", "IsAlone"]].head())

print("\n---\n")

# 3. Create pivot table
print("3. Pivot Table - Survival Rate by Sex and Pclass")

pivot_table = df.pivot_table(
    values="Survived",
    index="Sex",
    columns="Pclass",
    aggfunc="mean"
)

display(pivot_table)

print("\n---\n")

# 4. Identify highest and lowest survival combinations
print("4. Highest and Lowest Survival Combinations")

highest = pivot_table.max().max()
lowest = pivot_table.min().min()

print("Highest survival rate:", highest)
print("Lowest survival rate:", lowest)

# Find corresponding combinations
highest_pos = pivot_table.stack().idxmax()
lowest_pos = pivot_table.stack().idxmin()

print("Highest combination (Sex, Pclass):", highest_pos)
print("Lowest combination (Sex, Pclass):", lowest_pos)
```
